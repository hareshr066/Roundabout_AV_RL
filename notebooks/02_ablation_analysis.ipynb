This notebook analyzes the contribution of each component in our RL framework.

## 2. Visualization

```python
import os
csv_path = '../results/ablation_study_results.csv'
if os.path.exists(csv_path):
    df = pd.read_csv(csv_path)
    fig, ax = plt.subplots(figsize=(12, 6))
    x = np.arange(len(df))
    width = 0.25
    ax.bar(x - width, df['success_rate'], width, label='Success Rate', color='#2ecc71')
    ax.bar(x, df['collision_rate'], width, label='Collision Rate', color='#e74c3c')
    ax.bar(x + width, df['timeout_rate'], width, label='Timeout Rate', color='#f39c12')
    ax.set_xticks(x)
    ax.set_xticklabels([v.split('. ')[1] if '. ' in v else v for v in df['Variant']], rotation=30, ha='right')
    ax.set_ylabel('Rate (%)')
    ax.set_title('Ablation Study: Component Contributions')
    ax.legend()
    plt.tight_layout()
    plt.savefig('../results/figures/ablation_notebook.png', dpi=150)
    plt.show()
```
