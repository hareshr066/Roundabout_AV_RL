# Training Analysis

This notebook analyzes training curves and curriculum progression from TensorBoard logs.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tensorboard.backend.event_processing import event_accumulator

plt.style.use('seaborn-v0_8-whitegrid')
print('Libraries loaded successfully!')
```

## 1. Load TensorBoard Logs

```python
# Load TensorBoard event files
log_dir = '../results/logs/tb'

def load_tb_events(log_dir, tag):
    """Load a specific metric from TensorBoard logs."""
    ea = event_accumulator.EventAccumulator(log_dir)
    ea.Reload()
    if tag in ea.Tags()['scalars']:
        events = ea.Scalars(tag)
        steps = [e.step for e in events]
        values = [e.value for e in events]
        return pd.DataFrame({'step': steps, tag: values})
    return pd.DataFrame()

print('Available log directories:', os.listdir(log_dir) if os.path.exists(log_dir) else 'Not found')
```

## 2. Plot Training Curves

```python
# Plot reward and success rate over training
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Placeholder - replace with actual TensorBoard data
steps = np.arange(0, 500000, 1000)
rewards = np.random.randn(len(steps)).cumsum() * 0.1
success_rates = np.clip(np.random.randn(len(steps)).cumsum() * 0.001 + 0.5, 0, 1)

axes[0].plot(steps, rewards, alpha=0.7)
axes[0].set_xlabel('Training Steps')
axes[0].set_ylabel('Episode Reward')
axes[0].set_title('Training Reward Curve')

axes[1].plot(steps, success_rates * 100, color='green', alpha=0.7)
axes[1].set_xlabel('Training Steps')
axes[1].set_ylabel('Success Rate (%)')
axes[1].set_title('Success Rate Over Training')

plt.tight_layout()
plt.savefig('../results/figures/training_curves_notebook.png', dpi=150)
plt.show()
print('Saved training curves figure.')
```
